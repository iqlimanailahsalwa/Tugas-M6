```python
#nomor 1

usia = int(input("Masukkan usia: "))
nama_departemen = input("Masukkan nama departemen: ")

hasil_seleksi = usia >=19 and "Statistika" in nama_departemen

print(f"Formulir yang memenuhi syarat : {hasil_seleksi}")
```

```python
#nomor 2

data_pelanggan = ("Salwa","Premium")
berat_kardus = [65,43,28,12]
nama,status_member = data_pelanggan
total_berat = sum(berat_kardus)
print(total_berat)
print(f"Pelanggan prioritas VIP : {status_member == "Premium" or total_berat > 100}")

data_pelanggan = ("Salwa","Not Premium")
berat_kardus = [65,59,28]
nama,status_member = data_pelanggan
total_berat = sum(berat_kardus)
print(total_berat)
print(f"Pelanggan prioritas VIP : {status_member == "Premium" or total_berat > 100}")
```
